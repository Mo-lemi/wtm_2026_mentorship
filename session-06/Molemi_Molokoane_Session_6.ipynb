### Question 1: First 10 records

```python
query = """
SELECT *
FROM orders
LIMIT 10
"""

pd.read_sql_query(query, conn)
```

**What the result tells us:** Each row is one order with its customer, status and the timestamps for purchase, approval, carrier hand-off, customer delivery and estimated delivery. The dates are stored as text, so they would need converting before doing date calculations.

### Question 2: Delivered orders

```python
query = """
SELECT *
FROM orders
WHERE order_status = 'delivered'
"""

pd.read_sql_query(query, conn)
```

**What the result tells us:** 96,478 orders have the status `delivered`, so the vast majority of orders reached the customer. A few of these (8) still have an empty `order_delivered_customer_date`, which is a data quality issue worth checking.

### Question 3: Total number of orders

```python
query = """
SELECT COUNT(*) AS total_orders
FROM orders
"""

pd.read_sql_query(query, conn)
```

**What the result tells us:** The `orders` table contains 99,441 orders in total. This matches the number of rows we loaded from the CSV, so no data was lost when moving from Pandas into SQLite.

### Question 4: Orders per status

```python
query = """
SELECT order_status, COUNT(*) AS total_orders
FROM orders
GROUP BY order_status
"""

pd.read_sql_query(query, conn)
```

**What the result tells us:** There are 8 different order statuses. Grouping shows how orders are spread across the fulfilment pipeline, from `created` and `approved` through to `delivered`, `canceled` and `unavailable`.

### Question 5: Statuses sorted from most to fewest orders

```python
query = """
SELECT order_status, COUNT(*) AS total_orders
FROM orders
GROUP BY order_status
ORDER BY total_orders DESC
"""

pd.read_sql_query(query, conn)
```

**What the result tells us:** `delivered` (96,478) is by far the most common status, followed by `shipped` (1,107), `canceled` (625) and `unavailable` (609). Only about 3% of orders are not delivered, but canceled and unavailable orders together (1,234) are lost sales worth investigating.
